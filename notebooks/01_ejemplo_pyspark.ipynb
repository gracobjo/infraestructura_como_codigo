# Ejemplo básico con PySpark

Este notebook se ejecuta en el contenedor `python-env` (imagen `jupyter/pyspark-notebook`).

Pasos típicos para la clase:
1. Crear una `SparkSession` conectada al **Spark Master** (`spark-master:7077`).
2. Crear un DataFrame en memoria.
3. Hacer una operación simple (count, groupBy, etc.).

Asegúrate de haber levantado el entorno con `docker compose up -d` antes de ejecutar las celdas.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("EjemploPySparkCodespaces") \
    .master("spark://spark-master:7077") \
    .getOrCreate()

spark
```

```python
datos = [
    ("Alice", 34),
    ("Bob", 45),
    ("Carmen", 29),
    ("Diego", 41)
]

df = spark.createDataFrame(datos, ["nombre", "edad"])
df.show()
```

```python
from pyspark.sql.functions import avg

df.groupBy().agg(avg("edad").alias("edad_media")).show()
```
